```python
import re

# 提供的文本内容
content = """
[CONTENT]
"Required packages": [
    "flask==1.1.2"
],
"Required Other language third-party packages": [
    "No third-party dependencies required"
],
"Logic Analysis": [
    ...
],
"Task list": [
    T0:|handle login part|implement user login functionality|[]|related files:["main.py", "templates/login.html"]
    T1:|handle registration part|implement user registration functionality|[T0]|related files:["main.py", "templates/register.html"]
    T2:|setup home page|implement home page functionality|[T1]|related files:["main.py", "templates/home.html"]
    T3:|recipe submission|implement recipe submission functionality|[T2]|related files:["main.py", "templates/submit_recipe.html"]
    T5:|recipe details|implement recipe details functionality|[T4]|related files:["main.py", "templates/recipe_details.html"]
],
"Shared Knowledge": "`main.py` contains the core application logic and routing for the web application.",
[/CONTENT]
"""

# 正则表达式模式，用于匹配 Task list 部分的内容
task_list_pattern = r'"Task list": \[(.*?)\],'
match = re.search(task_list_pattern, content, re.S)

if match:
    task_list_content = match.group(1).strip()
    cleaned_content = re.sub(r"^\s+", "", task_list_content, flags=re.MULTILINE)
    print(cleaned_content)
else:
    print("No Task List found.")
```

```python
# 创建字典
task_dict = {}
task_pattern = r"^(T\d+):\|(.*)$"
# 遍历每行内容并构建字典
for line in cleaned_content.strip().splitlines():
    match = re.match(task_pattern, line)
    if match:
        key = match.group(1)  # 提取任务ID
        value = match.group(2)  # 提取任务详情
        task_dict[key] = value

# 打印结果
print("Task Dictionary:")
print(task_dict)
```
